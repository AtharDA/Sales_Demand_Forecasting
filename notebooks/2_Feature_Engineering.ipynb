# Sales & Demand Forecasting
## Notebook 2: Feature Engineering

### Objective
Prepare weekly sales data for forecasting by removing the incomplete final week and creating useful date-based features.

### Scope
- Load the processed weekly sales data.
- Remove the incomplete final week.
- Create calendar features and a continuous time index.
- Save the prepared dataset for model building.

No forecasting models are trained in this notebook.


```python
import pandas as pd
from pathlib import Path

# Project paths
PROJECT_DIR = Path("..")
PROCESSED_DATA_DIR = PROJECT_DIR / "data" / "processed"

# Load weekly sales data
weekly_sales = pd.read_csv(
    PROCESSED_DATA_DIR / "weekly_sales.csv",
    parse_dates=["date"]
)

print("Original shape:", weekly_sales.shape)
print("Date range:", weekly_sales["date"].min(),
      "to", weekly_sales["date"].max())

display(weekly_sales.head())
display(weekly_sales.tail())
```

## 1. Handle the Incomplete Final Week

The dataset ends on August 15, 2017, so the final weekly aggregate is incomplete. Remove this record before preparing the dataset for forecasting.

```python
# Remove the incomplete final week
weekly_sales = weekly_sales.iloc[:-1].copy()

# Create date-based features
weekly_sales["year"] = weekly_sales["date"].dt.year
weekly_sales["month"] = weekly_sales["date"].dt.month
weekly_sales["week_of_year"] = (
    weekly_sales["date"].dt.isocalendar().week.astype(int)
)

# Continuous time index for modeling the long-term trend
weekly_sales["time_index"] = range(len(weekly_sales))

print("Shape after removing incomplete week:", weekly_sales.shape)
print("Date range:", weekly_sales["date"].min(),
      "to", weekly_sales["date"].max())

display(weekly_sales.head())
```

```python
# Save prepared data
output_path = PROCESSED_DATA_DIR / "weekly_sales_features.csv"

weekly_sales.to_csv(output_path, index=False)

print("Saved to:", output_path)
```

## 2. Feature Engineering Summary

- Removed the incomplete final weekly observation because the source data ends on August 15, 2017.
- Created calendar features: `year`, `month`, and `week_of_year`.
- Added `time_index`, a continuous integer that increases by one for each week. This lets Linear Regression represent the long-term trend without the week number resetting at the start of each year.
- Saved the prepared dataset as `data/processed/weekly_sales_features.csv`.
